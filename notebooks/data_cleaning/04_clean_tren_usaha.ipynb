# 03 Perhitungan Laju Pertumbuhan Tahunan YoY
---
## Analisis Pertumbuhan
---
### Kalkulasi Persentase Pertumbuhan YoY
---
Menghitung persentase perubahan volume unit usaha tiap tahun terhadap tahun sebelumnya.


```python
df_tren['growth_yoy_pct'] = df_tren.groupby('sektor')['jumlah_unit'].pct_change() * 100
df_tren['growth_yoy_pct'] = df_tren['growth_yoy_pct'].fillna(0.0)
df_tren.tail(10)
```
